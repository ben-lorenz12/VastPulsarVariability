```python
# importing required modules
import matplotlib.pyplot as plt
from vasttools.query import Query
import pandas as pd

from vasttools.moc import VASTMOCS
from astropy import units as u

import numpy as np

from astropy.coordinates import SkyCoord

import scipy.stats as stats
```

```python
ctrl_df = pd.read_csv('all_final_controls.csv')
```

```python
ctrl_df.drop(columns=ctrl_df.columns[:2], inplace=True)
ctrl_df.drop(columns=ctrl_df.columns[1:3], inplace=True)
```

```python
names = ['J0907-5157',
         'J0908-4913',
         'J0955-5304',
         'J1544-5308',
         'J1733-3716',
         'J1808-1020',
         'J1813-2621',
         'J1840-1207'
        ]
```

```python
for psr_name in names:    
    ctrl_mask = ctrl_df['PSR_assoc']==psr_name

    new_ctrls = pd.read_csv('manual_ctrl_lists/' + psr_name + '.csv')

    new_ctrls = new_ctrls[new_ctrls['num_detections']>39]
    new_ctrls.insert(0, 'PSR_assoc', psr_name)
    new_ctrls.drop(columns=new_ctrls.columns[1:3], inplace=True)

    ctrl_df.drop(
        labels = ctrl_df[ctrl_mask].index,
        inplace=True
    )

    ctrl_df = pd.concat([ctrl_df, new_ctrls], ignore_index=True)
```

```python
ctrl_df.to_csv('all_final_ctrls.csv')
```
